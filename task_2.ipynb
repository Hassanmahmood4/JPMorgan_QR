## 1. Contract Overview

A storage contract lets you:
- **Inject** gas on chosen dates (buy at market price, store it)
- **Withdraw** gas on chosen dates (sell at market price)

**Assumptions (per task brief):**
- No transport delay — gas moves instantly
- Zero interest rates — no discounting
- Weekends and holidays are ignored

**Contract value** = withdrawal revenue − injection cost − storage fees
