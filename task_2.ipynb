## 1. Contract Overview

A storage contract lets you:
- **Inject** gas on chosen dates (buy at market price, store it)
- **Withdraw** gas on chosen dates (sell at market price)

**Assumptions (per task brief):**
- No transport delay — gas moves instantly
- Zero interest rates — no discounting
- Weekends and holidays are ignored

**Contract value** = withdrawal revenue − injection cost − storage fees

## 2. Pricing Function

The function takes six inputs:
1. Injection dates
2. Withdrawal dates
3. Prices (dict or `get_price_estimate` from Task 1)
4. Injection/withdrawal rate (max volume per event)
5. Maximum storage volume
6. Storage cost (per unit per day)

```python
from storage_contract import calculate_contract_value


def price_storage_contract(
    injection_dates,
    withdrawal_dates,
    prices,
    rate,
    max_volume,
    storage_cost,
) -> float:
    """Return the total value of a gas storage contract."""
    return calculate_contract_value(
        injection_dates=injection_dates,
        withdrawal_dates=withdrawal_dates,
        prices=prices,
        rate=rate,
        max_volume=max_volume,
        storage_cost=storage_cost,
    )
```
